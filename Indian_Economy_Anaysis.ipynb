```python
import numpy as np

i["GNP"] = (np.array(i["GDP_Lakh_Crore"]) + (np.array(i["Imports_Crore"]) - np.array(i["Exports_Crore"])) / 10000)
```

```python
i
```

```python
i.GNP
```

```python
i.groupby("Year")["GDP_Lakh_Crore"].sum()
```

```python
i
```

```python
i.groupby("Quarter")["GDP_Lakh_Crore"].sum()
```

```python
i.groupby("Year")[["GDP_Lakh_Crore", "GNP"]].sum().plot(
    kind="line",
    title="Year-wise GDP and GNP"
)

plt.xlabel("Year")
plt.ylabel("GDP and GNP")
plt.grid(True)
plt.show()
```

```python
i.groupby("Quarter")[["GDP_Lakh_Crore", "GNP"]].sum().plot(
    kind="bar",
    title="Year-wise GDP and GNP"
)

plt.xlabel("Year")
plt.ylabel("GDP and GNP")
plt.grid(True)
plt.show()
```
